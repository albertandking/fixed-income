## 编程实验 5：环境自检 + 画第一张收益率曲线


```python
import sys, numpy, pandas, scipy, fi
print('Python', sys.version.split()[0], '| fi', fi.__version__)
print('numpy', numpy.__version__, '| pandas', pandas.__version__, '| scipy', scipy.__version__)
try:
    import QuantLib; print('QuantLib', QuantLib.__version__)
except ImportError:
    print('QuantLib 未安装（uv sync --extra all 安装）')

from fi import data, plotting
plotting.use_chinese_style()
curve = data.load_sample('cgb_yield_curve')
fig, ax = plotting.new_axes()
ax.plot(curve['tenor'], curve['yield_pct'], marker='o')
ax.set_xlabel('期限（年）'); ax.set_ylabel('到期收益率 (%)'); ax.set_title('中国国债收益率曲线（样本）')
fig.tight_layout()
```

## 编程实验 6：叠加 +50bp 平行移动


```python
fig, ax = plotting.new_axes()
ax.plot(curve['tenor'], curve['yield_pct'], marker='o', label='基准曲线')
ax.plot(curve['tenor'], curve['yield_pct'] + 0.50, marker='s', label='整体上行 50bp')
ax.set_xlabel('期限（年）'); ax.set_ylabel('到期收益率 (%)'); ax.legend(); ax.set_title('收益率曲线平行移动')
fig.tight_layout()
print('平行移动：两条线上下平移、形状不变')
```
