```python
import numpy as np
from fi import rateopt as ro
from fi import plotting
plotting.use_chinese_style()
resets, taus = [1, 2, 3, 4], [1, 1, 1, 1]
pay_dfs = [1.03 ** -t for t in (2, 3, 4, 5)]
fwds = [0.03] * 4; M = 1e8; vol = 0.20
```

## 例13.1　单个 caplet（重置 1y、支付 2y）


```python
cl = ro.black_caplet(0.03, 0.03, vol, 1, 1, pay_dfs[0], M, 'cap')
print(f'caplet 价值 = {cl:,.0f} 元')
```

## 例13.2　ATM Cap / Floor 与 Cap-Floor 平价


```python
for K in (0.02, 0.03, 0.04):
    cap = ro.black_cap(fwds, K, vol, resets, taus, pay_dfs, M, 'cap')
    flr = ro.black_cap(fwds, K, vol, resets, taus, pay_dfs, M, 'floor')
    swp = M * (sum(fwds[i]*taus[i]*pay_dfs[i] for i in range(4)) - K*sum(pay_dfs))
    print(f'K={K*100:.0f}%: Cap={cap:>12,.0f}  Floor={flr:>12,.0f}  Cap-Floor={cap-flr:>12,.0f}  payerSwap={swp:>12,.0f}')
```

## 图13-1　Cap/Floor 价值随执行价（编程实验 7）


```python
strikes = np.linspace(0.015, 0.045, 31)
caps = [ro.black_cap(fwds, K, vol, resets, taus, pay_dfs, M, 'cap')/1e4 for K in strikes]
floors = [ro.black_cap(fwds, K, vol, resets, taus, pay_dfs, M, 'floor')/1e4 for K in strikes]
fig, ax = plotting.new_axes()
ax.plot(strikes*100, caps, label='Cap 价值'); ax.plot(strikes*100, floors, label='Floor 价值')
ax.axvline(3.0, ls=':', color='gray', label='ATM (3%)')
ax.set_xlabel('执行利率 K (%)'); ax.set_ylabel('价值（万元）')
ax.set_title('图13-1　Cap 与 Floor 价值随执行价'); ax.legend()
fig.tight_layout()
```

## 例13.3　Swaption（1y -> 4y payer）


```python
sp = ro.black_swaption(0.03, 0.03, vol, expiry=1, swap_annuity=sum(pay_dfs), notional=M, kind='payer')
print(f'payer swaption 价值 = {sp:,.0f} 元')
```

## 例13.4　隐含波动率 + 波动率微笑（编程实验 8）


```python
iv = ro.implied_vol(cl, 0.03, 0.03, 1, 1, pay_dfs[0], M, 'cap')
print(f'由 caplet 价格反求隐含波动率 = {iv*100:.2f}%（应=20%）')

# 合成一个带偏斜的波动率，造价格，再反求 -> 画微笑
strikes2 = np.linspace(0.02, 0.04, 9)
true_vol = lambda K: 0.20 + 4.0*(K-0.03)**2/0.0001*0.002   # 人造微笑
prices = [ro.black_caplet(0.03, K, true_vol(K), 1, 1, pay_dfs[0], M, 'cap') for K in strikes2]
ivs = [ro.implied_vol(p, 0.03, K, 1, 1, pay_dfs[0], M, 'cap') for p, K in zip(prices, strikes2)]
fig, ax = plotting.new_axes()
ax.plot(strikes2*100, np.array(ivs)*100, marker='o')
ax.set_xlabel('执行利率 K (%)'); ax.set_ylabel('隐含波动率 (%)')
ax.set_title('波动率微笑（合成示例）')
fig.tight_layout()
```

## 13.7　QuantLib Cap 对拍（编程实验 9）


```python
import math
import QuantLib as ql
today = ql.Date(15, 6, 2026); ql.Settings.instance().evaluationDate = today
dc, cal = ql.Actual365Fixed(), ql.NullCalendar()
# 连续复利 log(1.03) 对应年度简单远期 3%，以便与本章示例对齐
ts = ql.YieldTermStructureHandle(ql.FlatForward(today, math.log(1.03), dc, ql.Continuous))
idx = ql.IborIndex('Idx', ql.Period(1, ql.Years), 0, ql.CNYCurrency(), cal,
                   ql.Unadjusted, False, dc, ts)
sched = ql.Schedule(today, today + ql.Period(5, ql.Years), ql.Period(1, ql.Years), cal,
                    ql.Unadjusted, ql.Unadjusted, ql.DateGeneration.Forward, False)
cap = ql.Cap(ql.IborLeg([1e8], sched, idx), [0.03])
cap.setPricingEngine(ql.BlackCapFloorEngine(ts, ql.QuoteHandle(ql.SimpleQuote(0.20)), dc))
fi_cap = ro.black_cap(fwds, 0.03, vol, resets, taus, pay_dfs, M, 'cap')
print(f'fi.rateopt ATM Cap = {fi_cap:,.0f} 元')
print(f'QuantLib   ATM Cap = {cap.NPV():,.0f} 元（统一复利方式后，剩余差异来自第2期跨2028闰年的计息天数）')
```

---

> 小结：利率期权用 Black 模型定价，caplet 到期=重置日、折现用支付日；Cap-Floor=payer 互换；
> 隐含波动率与波动率曲面是交易核心；可赎回债在简化框架下可近似理解为内嵌 receiver swaption。
