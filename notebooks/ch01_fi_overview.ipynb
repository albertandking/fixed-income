## 例1.1　一只国债的现金流

面值 100、票息 3%、年付息、剩余 3 年。用 `fi.cashflow.make_cashflows` 生成约定现金流。


```python
import pandas as pd
from fi.cashflow import make_cashflows
cfs, ts = make_cashflows(coupon_rate=0.03, maturity=3, freq=1, face=100)
tbl = pd.DataFrame({'时间(年)': ts, '现金流': cfs})
tbl['构成'] = ['票息', '票息', '票息 + 本金']
print(tbl.to_string(index=False))
```

## 1.7　本书第一张图：中国国债收益率曲线


```python
from fi import data, plotting
curve = data.load_sample('cgb_yield_curve')
print(curve.to_string(index=False))

plotting.use_chinese_style()
fig, ax = plotting.new_axes()
ax.plot(curve['tenor'], curve['yield_pct'], marker='o')
ax.set_xlabel('期限（年）'); ax.set_ylabel('到期收益率 (%)')
ax.set_title('图1-1　中国国债收益率曲线（样本数据）')
fig.tight_layout()
```

### 叠加平行移动 +50bp（编程实验 6）


```python
fig, ax = plotting.new_axes()
ax.plot(curve['tenor'], curve['yield_pct'], marker='o', label='基准曲线')
ax.plot(curve['tenor'], curve['yield_pct'] + 0.50, marker='s', label='整体上行 50bp')
ax.set_xlabel('期限（年）'); ax.set_ylabel('到期收益率 (%)')
ax.set_title('收益率曲线的平行移动'); ax.legend()
fig.tight_layout()
```
